# Notebook 04: Comparative Model Evaluation

Results produced by `python scripts/evaluate_model.py --dataset mavecodd`: both models on the same test tiles, train/val/test overfitting check, turbidity stress test and cross-domain transfer.


```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import json
import pandas as pd
import matplotlib.pyplot as plt

with open('../outputs/metrics/test_metrics_mavecodd.json', 'r') as f:
    metrics = json.load(f)
print(metrics['evaluation_protocol']['split_sizes'])
```

## 1. Overfitting check (train vs held-out)


```python
pd.DataFrame(metrics['overfitting_check']).T
```

## 2. Model A vs Model B on the same test tiles


```python
base = metrics['baseline_model_a']['test']
unet = metrics['deep_learning_model_b']['test']['patch_classification']
keys = ['balanced_accuracy', 'accuracy', 'precision', 'recall', 'f1_score', 'roc_auc']
pd.DataFrame({'Model A (features+SVM)': [base[k] for k in keys], 'Model B (U-Net)': [unet[k] for k in keys]}, index=keys)
```

## 3. Optical turbidity stress test


```python
deg = pd.DataFrame(metrics['optical_degradation_stress_test'])
plt.figure(figsize=(8, 4.5))
plt.plot(deg['turbidity'], deg['unet_mean_dice'], marker='o', label='U-Net Dice')
plt.plot(deg['turbidity'], deg['unet_patch_balanced_accuracy'], marker='s', label='U-Net tile balanced acc')
plt.plot(deg['turbidity'], deg['baseline_patch_balanced_accuracy'], marker='^', label='SVM tile balanced acc')
plt.xlabel('Turbidity')
plt.ylim(0, 1.05)
plt.legend()
plt.grid(True)
plt.show()
```
